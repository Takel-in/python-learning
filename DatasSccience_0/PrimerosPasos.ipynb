6 IF
if <condicion>:
    <code>
elif <condicion>:
    <code>
----
else :
    <code>

```python
edad = 24

if edad >= 18:
    print("Eres mayor de edad")
elif edad >= 13:
    print( "Eres adolescente")
else:
    print("No eres mayor de edad")

```

```python
countries = ['United States', 'India', 'China', 'Brasil']
if 'Colombia' in countries:
    print ("Pais está en la lista")
else:
    print ("País NO está en lal ista")
```

7 For Loop (bucle for)
for <variable> in <list>:
    <code>

```python
countries = ['United States', 'India', 'China', 'Brasil']
for pais in countries:
    print (pais)
```

```python
for pais in countries:
    if pais == 'United States':
        print(pais)
```

```python
for cnt, pais in enumerate(countries):
    print(cnt)
    print(pais)
```

```python
my_data = {'nombre':'Frank','Edad':26}
my_data
```

```python
my_data.items()
```

```python
for clave, valor in my_data.items():
    print (clave)
    print (valor)
    print (clave, valor)
```
