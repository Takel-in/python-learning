1 CREANDO UN DATAFRAME USANDO ARRAYS

1.1 Opcion1 Arrays con numpy

```python
#Creamos un Array con np.
data = np.array([[1,4], [2,5], [3,4]]) # cada conjunto es una fila.
```

```python
#Crear un dataframe
df = pd.DataFrame(data, 
             index=['row1', 'row2', 'row3'],
             columns=['col1', 'col2'] )
df
```

1.2 Opcion1 Arrays sin numpy

```python
# El array 
data = [[1,4], [2,5], [3,4]]  
```

```python
#Crear un dataframe
df = pd.DataFrame(data, 
             index=['row1', 'row2', 'row3'],
             columns=['col1', 'col2'] )
df
```
